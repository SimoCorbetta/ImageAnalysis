For segmentation, we used the algorithm otsu threshold after converting the image to grey tones.
This threshold is determined by minimizing intra-class intensity variance, or equivalently, by maximizing inter-class variance.
We then removed objects that were touching the border of the image and we only labelled objects that have a size bigger than 150
to be sure to label only cells and eventually not noisy pixels

```python
def segmentation(image):
    image_g = dip.ColorSpaceManager.Convert(image,"grey")
    image_g.SetPixelSize(1, "um")
    otsu_image_g = dip.OtsuThreshold(image_g)
    otsu_image_g = dip.EdgeObjectsRemove(otsu_image_g)
    otsu_image_g = dip.Label(otsu_image_g, minSize=150)
    
    m = dip.MeasurementTool.Measure(otsu_image_g,image_g ,measure)
    return m
```

function to access the x and y coordinates of center of mass of each object and store them in a dataframe where rows
are the objects found in the images, and columns are the succession of frames

```python
def get_center(l, n):
    X = pd.DataFrame()
    Y = pd.DataFrame()
    
    for i in range(30):
        c = l[i]['Center']
        c = np.array(c)[:n,]
        df_c = pd.DataFrame(c, columns=['x', 'y'])
        X[f'frame{i}'] = df_c['x'].reset_index(drop=True)
        Y[f'frame{i}'] = df_c['y'].reset_index(drop=True)
    
    return X, Y
```

```python
def get_features(l, n):
    size = pd.DataFrame()
    perim = pd.DataFrame()
    roundness = pd.DataFrame()
    mean = pd.DataFrame()
    std = pd.DataFrame()
    maxval = pd.DataFrame()
    
    for i in range(30):
        size_i = l[i]['Size']
        perim_i = l[i]['Perimeter']
        roundness_i = l[i]['Roundness']
        mean_i = l[i]['Mean']
        std_i = l[i]['StandardDeviation']
        maxval_i = l[i]['MaxVal']
        size_i = np.array(size_i)[:n]
        perim_i = np.array(perim_i)[:n]
        roundness_i = np.array(roundness_i)[:n]
        mean_i = np.array(mean_i)[:n]
        std_i = np.array(std_i)[:n]
        maxval_i = np.array(maxval_i)[:n]
        size_df = pd.DataFrame(size_i, columns=['size'])
        perim_df = pd.DataFrame(perim_i, columns=['perim'])
        roundness_df = pd.DataFrame(roundness_i, columns=['roundness'])
        mean_df = pd.DataFrame(mean_i, columns=['mean'])
        std_df = pd.DataFrame(std_i, columns=['std'])
        maxval_df = pd.DataFrame(maxval_i, columns=['maxval'])
        
        size[f'frame{i}'] = size_df['size'].reset_index(drop=True)
        perim[f'frame{i}'] = perim_df['perim'].reset_index(drop=True)
        roundness[f'frame{i}'] = roundness_df['roundness'].reset_index(drop=True)
        mean[f'frame{i}'] = mean_df['mean'].reset_index(drop=True)
        std[f'frame{i}'] = std_df['std'].reset_index(drop=True)
        maxval[f'frame{i}'] = maxval_df['maxval'].reset_index(drop=True)
        
    return size, perim, roundness, mean, std, maxval
```

I use nearest object as criteria to identify the object in the next frame.
The validity of this criteria relies on the assumption that cells are sparsely scattered and their traveling speeds between one frame and the other are relatively uniform and of the same magnitude.
Attention: this assumption sometimes is wrong, but it holds for vast majority of cases cause cell coltures tend to be pretty 
homogenous if the growing conditions do not vary, as it is this case.

```python
def tracing(X, Y, size, perim, roundness, mean, std, maxval,n):
    trace_x = pd.DataFrame()
    trace_y = pd.DataFrame()
    trace_label = pd.DataFrame()

    trace_size = pd.DataFrame()
    trace_perim = pd.DataFrame()
    trace_roundness = pd.DataFrame()
    trace_mean = pd.DataFrame()
    trace_std = pd.DataFrame()
    trace_maxval = pd.DataFrame()
    trace_dist = pd.DataFrame()

    trace_x['frame0'] = X['frame0']
    trace_y['frame0'] = Y['frame0']
    trace_label['frame0'] = np.arange(n)+1
    trace_size['frame0'] = size['frame0']
    trace_perim['frame0'] = perim['frame0']
    trace_roundness['frame0'] = roundness['frame0']
    trace_mean['frame0'] = mean['frame0']
    trace_std['frame0'] = std['frame0']
    trace_maxval['frame0'] = maxval['frame0']
    for frame in range(29):
        # check out the current and next frame
        x_frame = np.array(trace_x.iloc[:,frame])
        y_frame = np.array(trace_y.iloc[:,frame])
        x_nextframe = np.array(X.iloc[:,frame+1])
        y_nextframe = np.array(Y.iloc[:,frame+1])
        size_nextframe = np.array(size.iloc[:,frame+1])
        perim_nextframe = np.array(perim.iloc[:,frame+1])
        round_nextframe = np.array(roundness.iloc[:,frame+1])
        mean_nextframe = np.array(mean.iloc[:,frame+1])
        std_nextframe = np.array(std.iloc[:,frame+1])
        maxval_nextframe = np.array(maxval.iloc[:,frame+1])
        
        X_next = []
        Y_next = []
        label_next = []
        size_next = []
        perim_next = []
        roundness_next = []
        mean_next = []
        std_next = []
        maxval_next = []
        tot_distance = []
        # loop through each label
        for l in range(n):
            # distance array 
            dist_i = ((x_frame[l] - x_nextframe)**2 + (y_frame[l] - y_nextframe)**2)**0.5
            # find the index of object in next frame closest to current label 
            i_next_match = np.argmin(dist_i)
            X_next.append(x_nextframe[i_next_match])
            Y_next.append(y_nextframe[i_next_match])
            label_next.append(i_next_match+1)
            size_next.append(size_nextframe[i_next_match])
            perim_next.append(perim_nextframe[i_next_match])
            roundness_next.append(round_nextframe[i_next_match])
            mean_next.append(mean_nextframe[i_next_match])
            std_next.append(std_nextframe[i_next_match])
            maxval_next.append(maxval_nextframe[i_next_match])
            tot_distance.append(dist_i[i_next_match])
        trace_x['frame{}'.format(frame+1)] = np.array(X_next)
        trace_y['frame{}'.format(frame+1)] = np.array(Y_next)
        trace_label['frame{}'.format(frame+1)] = np.array(label_next)
        trace_size['frame{}'.format(frame+1)] = np.array(size_next)
        trace_perim['frame{}'.format(frame+1)] = np.array(perim_next)
        trace_roundness['frame{}'.format(frame+1)] = np.array(roundness_next)
        trace_mean['frame{}'.format(frame+1)] = np.array(mean_next)
        trace_std['frame{}'.format(frame+1)] = np.array(std_next)
        trace_maxval['frame{}'.format(frame+1)] = np.array(maxval_next)
        trace_dist['frame{}to{}'.format(frame, frame+1)] = np.array(tot_distance)
    return trace_x, trace_y, trace_label, trace_size, trace_perim, trace_roundness, trace_mean, trace_std, trace_maxval, trace_dist
```

```python
trace_x_ctrl,trace_y_ctrl, trace_label_ctrl, trace_size_ctrl, trace_perim_ctrl, trace_roundness_ctrl, trace_mean_ctrl, trace_std_ctrl, trace_maxval_ctrl, trace_dist_ctrl = tracing(X_ctrl,Y_ctrl,size_ctrl,perim_ctrl,roundness_ctrl,mean_ctrl,std_ctrl,maxval_ctrl,min_n_ctrl)
```
